# diddesign: Double Difference-in-Differences for Python

This notebook demonstrates the complete workflow of the `diddesign` package, 
implementing Egami & Yamauchi (2023) Double DID method.

```python
from diddesign import did, did_check
from diddesign.data import data
from diddesign.visualization import plot_diagnostics_panel, plot_estimates
```

## Example 1: Repeated Cross-Section (RCS) Design

Using Malesky et al. (2014) data on Vietnamese provincial governance reforms.

```python
df = data("malesky2014")
print(f"Dataset: {df.shape[0]} observations, {df.shape[1]} variables")
print(f"Time periods: {sorted(df['year'].unique())}")
print(f"Treatment rate: {df['treatment'].mean():.1%}")

check = did_check(data=df, outcome="pro4", treatment="treatment",
                  time="year", post="post_treat", data_type="rcs",
                  id_cluster="id_district", lag=[1], n_boot=50, random_seed=1234)
check
```

```python
fig = plot_diagnostics_panel(check, show=True)
```

```python
result = did(df, outcome="pro4", treatment="treatment",
             time="year", post="post_treat", data_type="rcs",
             id_cluster="id_district", n_boot=100, random_seed=1234)
result
```

```python
result.to_dataframe()
```

```python
print(result.to_latex(caption="Double DID Estimates: Malesky (2014)", 
                      label="tab:malesky"))
```

## Example 2: Staggered Adoption (SA) Design

Using Paglayan (2019) data on collective bargaining laws and education spending.

```python
df_sa = data("paglayan2019")
result_sa = did(df_sa, outcome="pupil_expenditure", treatment="treatment",
                time="year", unit_id="state", design="sa",
                thres=1, n_boot=100, random_seed=1234)
result_sa
```

```python
result_sa.to_dataframe()
```
